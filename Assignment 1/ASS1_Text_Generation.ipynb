```python
import re

# 1. Define a string variable to hold your input text data.
text_data = """
Once upon a time, there lived a man in a small town. The man was a miser. He used to collect all his earnings and spent very little. He loved his gold more than anything else. So he decided and buried all his gold in a secret spot in his garden.

Each day, he would dig it up, count it, and then bury it again. One day, as per his daily ritual, the miser man was counting his gold, a thief saw everything. That same night the thief snuck into the garden, dug up the gold, and took it away. The miser was devastated. He started crying aloud.

Upon hearing his cries, the miser’s neighbour enquired, “What happened? why are you crying so loud? The miser told the neighbour that his golden treasure has been stolen.

“Why didn’t you keep the gold inside the house? It would have been easier for you to access it when you have to buy something, and it would have been ever difficult for the thief to steal it,” the neighbour asked.

“Buy something? I never use my money to buy anything. And I was never going to spend the gold that I had saved. The neighbour nodded his head in disbelief and replied, “If that was the case then probably you should’ve saved the stones instead, the gold was not worth saving anyway.”"""
print(f"Original text: {text_data}")

# 2. Convert the entire text to lowercase.
text_data_lower = text_data.lower()
print(f"Lowercase text: {text_data_lower}")

# 3. Remove any punctuation marks from the text.
# Using a regular expression to replace non-alphanumeric characters (except spaces) with an empty string.
text_data_no_punct = re.sub(r'[^a-z0-9\s]', '', text_data_lower)
print(f"Text without punctuation: {text_data_no_punct}")

# 4. Split the cleaned text into a list of individual words.
words = text_data_no_punct.split()
print(f"Tokenized words: {words}")
```

```python
markov_chain = {}

for i in range(len(words) - 1):
    current_word = words[i]
    next_word = words[i+1]

    if current_word not in markov_chain:
        markov_chain[current_word] = {}

    if next_word not in markov_chain[current_word]:
        markov_chain[current_word][next_word] = 1
    else:
        markov_chain[current_word][next_word] += 1

print("Markov Chain Model:")
print(markov_chain)
```

```python
import random

def generate_text(chain, length):
    # 3. Select a random starting word from the keys of the markov_chain.
    current_word = random.choice(list(chain.keys()))

    # 4. Initialize a list to store the generated words, starting with the chosen initial word.
    generated_words = [current_word]

    # 5. Loop length - 1 times to generate the subsequent words:
    for _ in range(length - 1):
        # a. Get the current word (the last word added to your generated list).
        current_word = generated_words[-1]

        # b. Look up the possible next words and their frequencies from the markov_chain using the current word.
        if current_word in chain:
            next_word_candidates = chain[current_word]

            # Extract words and counts for random.choices
            words = list(next_word_candidates.keys())
            weights = list(next_word_candidates.values())

            # c. If there are no possible next words for the current word, select a new random starting word.
            if not words:
                current_word = random.choice(list(chain.keys()))
                generated_words.append(current_word)
                continue # Skip to the next iteration to find a successor for the new word

            # d. Use random.choices to select the next word based on the frequencies of the possible next words.
            next_word = random.choices(words, weights=weights, k=1)[0]

        else:
            # If the current_word is not in the chain (e.g., it was the last word in the original text),
            # select a new random starting word.
            next_word = random.choice(list(chain.keys()))

        # e. Append the chosen next word to your list of generated words.
        generated_words.append(next_word)

    # 6. After the loop, join the list of generated words with spaces to form a single string.
    return ' '.join(generated_words)

# 7. Call the generate_text function with markov_chain and a desired length (e.g., 20 words) and store the result in a variable.
generated_text_output = generate_text(markov_chain, 20)

# 8. Print the generated text.
print(f"Generated text (20 words): {generated_text_output}")
```

```python
nodes = []
edges = []

for current_word, next_word_counts in markov_chain.items():
    if current_word not in nodes:
        nodes.append(current_word)

    for next_word, count in next_word_counts.items():
        if next_word not in nodes:
            nodes.append(next_word)
        edges.append((current_word, next_word, count))

nodes = list(set(nodes))

print(f"First 10 Nodes: {nodes[:10]}")
print(f"First 10 Edges: {edges[:10]}")
```

```python
import random

node_positions = {}

for node in nodes:
    # Assign random x, y, z coordinates between -10 and 10
    x = random.uniform(-10, 10)
    y = random.uniform(-10, 10)
    z = random.uniform(-10, 10)
    node_positions[node] = (x, y, z)

# Print the node_positions for the first few nodes
print("Node Positions (first 10):")
for i, (node, pos) in enumerate(node_positions.items()):
    if i >= 10:
        break
    print(f"{node}: {pos}")
```

```python
node_x = []
node_y = []
node_z = []
node_labels = []

for node, pos in node_positions.items():
    node_x.append(pos[0])
    node_y.append(pos[1])
    node_z.append(pos[2])
    node_labels.append(node)

print(f"First 5 node_x: {node_x[:5]}")
print(f"First 5 node_y: {node_y[:5]}")
print(f"First 5 node_z: {node_z[:5]}")
print(f"First 5 node_labels: {node_labels[:5]}")
```

```python
edge_x = []
edge_y = []
edge_z = []
edge_widths = []

for edge in edges:
    source_word, target_word, count = edge

    # Get coordinates for source and target words
    source_pos = node_positions[source_word]
    target_pos = node_positions[target_word]

    # Append source coordinates
    edge_x.append(source_pos[0])
    edge_y.append(source_pos[1])
    edge_z.append(source_pos[2])
    # Scale width for better visibility; count=1 will have width 1.0
    edge_widths.append(count * 0.5 + 0.5)

    # Append target coordinates
    edge_x.append(target_pos[0])
    edge_y.append(target_pos[1])
    edge_z.append(target_pos[2])
    edge_widths.append(count * 0.5 + 0.5)

    # Add a None to create a break in the line for Plotly
    edge_x.append(None)
    edge_y.append(None)
    edge_z.append(None)
    edge_widths.append(None)

print(f"First 10 edge_x values: {edge_x[:10]}")
print(f"First 10 edge_y values: {edge_y[:10]}")
print(f"First 10 edge_z values: {edge_z[:10]}")
print(f"First 10 edge_widths values: {edge_widths[:10]}")
```

```python
import plotly.graph_objects as go

traces = []

# Create traces for the edges (one trace per edge for dynamic width)
for source_word, target_word, count in edges:
    source_pos = node_positions[source_word]
    target_pos = node_positions[target_word]

    edge_trace = go.Scatter3d(
        x=[source_pos[0], target_pos[0]],
        y=[source_pos[1], target_pos[1]],
        z=[source_pos[2], target_pos[2]],
        mode='lines',
        line=dict(color='#ADD8E6', width=count * 0.5 + 0.5), # Changed line color to light blue for better contrast
        hoverinfo='none',
        showlegend=False
    )
    traces.append(edge_trace)

# Create a trace for the nodes
node_trace = go.Scatter3d(
    x=node_x,
    y=node_y,
    z=node_z,
    mode='markers+text', # Show text directly
    marker=dict(
        symbol='circle',
        size=10, # Increased node size
        color='#1f77b4', # Changed node color to a more distinct blue
        line=dict(color='black', width=1) # Changed outline color to black and increased width
    ),
    text=node_labels,
    textposition='top center',
    textfont=dict(size=10, color='black'), # Set font size and color for node labels
    hoverinfo='text'
)
traces.append(node_trace)

# Create the 3D figure
fig = go.Figure(data=traces)

# Update layout for better visualization
fig.update_layout(
    title='3D Markov Chain Visualization',
    scene=dict(
        xaxis=dict(showbackground=False, showticklabels=False, showgrid=False, zeroline=False),
        yaxis=dict(showbackground=False, showticklabels=False, showgrid=False, zeroline=False),
        zaxis=dict(showbackground=False, showticklabels=False, showgrid=False, zeroline=False)
    ),
    showlegend=False,
    margin=dict(l=0, r=0, b=0, t=40)
)

# Display the figure
fig.show()
```
